```python
import math
x0,y0 = map(float,input().split())
R=float(input("R = "))
x,y=map(float,input("Tọa độ điểm A ").split())
d = math.sqrt((x-x0)**2+(y-y0)**2)
if d<R:
    print("Điểm A nằm trong đường tròn")
elif d==R:
    print("Điểm A nằm trên đường tròn")
else:
    if abs(d-R)<1e-9:
        print("Điểm A nằm trên đường tròn")
    else:
        print("Điểm A nằm ngoài đường tròn")
```
