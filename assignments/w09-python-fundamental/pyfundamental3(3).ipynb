## Problem 27

```python
import math
class Circle:
    def __init__(self,r):
        self.r=r
    def S(self):
        return math.pi*self.r**2;
r=float(input("radius="))
print(f"Area of circle: {Circle(r).S():.2f}")
```

## Problem 28

```python
class Rectangle:
    def __init__(self,a,b):
        self.a=a
        self.b=b
    def S(self):
        return self.a*self.b;
a,b=map(float,input("Enter the size:").split())
print(f"Area of rectangle: {Rectangle(a,b).S():.2f}")
```
